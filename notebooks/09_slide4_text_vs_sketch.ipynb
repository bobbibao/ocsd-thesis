# Slide 4: text-only vs sketch + text (real SD 1.5 images)

Makes the images for the 3-click animation on slide 4: the deck's sketch (duck, house, cow) and its caption **"a duck, a white house and a cow by a lake"** given (1) as text only to SD 1.5, and (2) with a sketch through SD 1.5 + ControlNet scribble (the same models OCSD and the ControlNet baseline use). 8 seeds each, so the best example can be picked.

**How to run:** Runtime -> Change runtime type -> any GPU (T4, L4 or A100) -> **Run all**. About 5 minutes. One ZIP downloads at the end; it is also on Drive at `KLTN_OCSD/slide4/SLIDE4_text_vs_sketch.zip`.

```
#@title 1. Drive, code, libraries
import os, subprocess, sys, shutil
from google.colab import drive
drive.mount("/content/drive")
BRANCH = "claude/project-thread-srtlwl"  #@param {type:"string"}
OUT = "/content/drive/MyDrive/KLTN_OCSD/slide4"
CODE = "/content/KLTN_OCSD_slide4"
shutil.rmtree(CODE, ignore_errors=True)
subprocess.run(["git", "clone", "-q", "-b", BRANCH, "https://github.com/bobbibao/ocsd-thesis.git", CODE], check=True)
os.environ["HF_HOME"] = "/content/hf_cache"
subprocess.run([sys.executable, "-m", "pip", "uninstall", "-y", "-q", "torchao"], check=True)
r = subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", os.path.join(CODE, "requirements-colab.txt")], capture_output=True, text=True)
print(r.stderr[-1500:])
if r.returncode:
    raise RuntimeError("pip install failed (see the text above)")
os.makedirs(OUT, exist_ok=True)
```

```
#@title 2. Text only vs sketch + text, 6 seeds each
import torch
from PIL import Image, ImageOps
from diffusers import StableDiffusionPipeline, StableDiffusionControlNetPipeline, ControlNetModel
PROMPT = "a duck, a white house and a cow by a lake"
NEG = "lowres, blurry, deformed, bad anatomy, extra limbs, watermark, text"
SEEDS = range(8)
sd = StableDiffusionPipeline.from_pretrained("stable-diffusion-v1-5/stable-diffusion-v1-5", torch_dtype=torch.float16, safety_checker=None).to("cuda")
cn = ControlNetModel.from_pretrained("lllyasviel/control_v11p_sd15_scribble", torch_dtype=torch.float16)
sdc = StableDiffusionControlNetPipeline(**sd.components, controlnet=cn).to("cuda")
sketch = Image.open(os.path.join(CODE, "notebooks/assets/slide4_sketch.png")).convert("RGB").resize((512, 512))
sketch.save(os.path.join(OUT, "sketch.png"))
cond = ImageOps.invert(sketch)  # scribble ControlNet wants white strokes on black
for s in SEEDS:
    g = torch.Generator("cuda").manual_seed(s)
    sd(PROMPT, negative_prompt=NEG, num_inference_steps=30, guidance_scale=7.5, generator=g).images[0].save(os.path.join(OUT, f"text_only_s{s}.png"))
    g = torch.Generator("cuda").manual_seed(s)
    sdc(PROMPT, image=cond, negative_prompt=NEG, num_inference_steps=30, guidance_scale=7.5, generator=g).images[0].save(os.path.join(OUT, f"sketch_text_s{s}.png"))
    print("seed", s, "done")
```

```
#@title 3. Sheet + ZIP, then download
import zipfile, glob
from IPython.display import display
rows = [[Image.open(os.path.join(OUT, f"{k}_s{s}.png")).resize((256, 256)) for s in SEEDS] for k in ("text_only", "sketch_text")]
sheet = Image.new("RGB", (256 * 8, 512), "white")
for i, row in enumerate(rows):
    for j, im in enumerate(row): sheet.paste(im, (256 * j, 256 * i))
sheet.save(os.path.join(OUT, "SHEET.jpg")); display(sheet)
zpath = os.path.join(OUT, "SLIDE4_text_vs_sketch.zip")
with zipfile.ZipFile(zpath, "w") as z:
    for p in glob.glob(os.path.join(OUT, "*.png")) + [os.path.join(OUT, "SHEET.jpg")]:
        z.write(p, os.path.basename(p))
print("ZIP:", zpath)
from google.colab import files
try:
    files.download(zpath)
except Exception as exc:
    print("Get it from Drive:", zpath, exc)
print("Done. Disconnect the GPU runtime.")
```
